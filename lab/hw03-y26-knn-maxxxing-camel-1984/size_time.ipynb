```python
from pathlib import Path
import pandas as pd
import metrics
from experiments import Model

path = Path("data/train.csv")
df = pd.read_csv(path)

X = df.drop(columns=["target"])
y = df["target"]
X_test = pd.read_csv("data/test.csv")

top = {"reduce": "top", "metric": "manhattan"}
```

```python
box = {**top, "n_features": 20, "n_objects": 12000, "k": 21}

size_kb, predict_time = metrics.size_and_time(Model, X, y, X_test, **box)
mean, std = metrics.mae(Model, X, y, **box)

print(f"размер: {size_kb:.1f} КБ (порог 1300, эталон 200)")
print(f"predict: {predict_time:.3f} с (порог 1.65, эталон 0.27)")
print(f"mae: {mean:.4f} ± {std:.4f} (порог 0.63, эталон 0.56)")
```

```python
params = {**top, "n_features": 20, "n_objects": 1000}
ks = [3, 5, 7, 9, 11, 15, 21, 31]

rows = []
for k in ks:
    mean, std = metrics.mae(Model, X, y, **params, k=k)
    rows.append({"k": k, "mae": mean, "std": std})

k_results = pd.DataFrame(rows)
display(k_results.round(4))
```

```python
n = [1000, 2000, 3000, 5000, 8000, 12000]
rows = []
for i in n:
    mean, std = metrics.mae(Model, X, y, **top, n_objects = i, n_features = 20)
    rows.append({"объектов": i, "mae": round(mean, 4), "std": round(std, 4)})
objects_results = pd.DataFrame(rows)
display(objects_results)
```
