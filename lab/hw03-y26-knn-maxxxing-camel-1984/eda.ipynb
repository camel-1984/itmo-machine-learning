```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

import eda
import visual

path = Path("data/train.csv")
df = pd.read_csv(path)
df.describe()
```

```python
classes = eda.target_distribution(df)
display(classes)
fig, ax = visual.target_distribution(classes)
plt.show()
```

```python
display(eda.feature_na(df))
uniq = eda.feature_nunique(df)
display(uniq)
fig, ax = visual.feature_nunique(uniq)
plt.show()
```

```python
fig, ax = visual.feature_std(eda.feature_std(df))
fig, ax = visual.feature_std_close_up(eda.feature_std(df))
plt.show()
```

```python
fig, ax = visual.feature_mean(eda.feature_mean(df))
fig, ax = visual.feature_mean_close_up(eda.feature_mean(df))
plt.show()
```

```python
eda.feature_strange(df)
```

```python
spearman = eda.feature_target_spearman(df.drop(columns="target"), df["target"])
fig, ax = visual.feature_target_spearman(spearman)
plt.show()
```

```python
mi = eda.feature_target_mi(df)
fig, ax = visual.feature_target_mi(mi)
plt.show()
```
