```python
import pandas as pd
import visualize


if __name__ == "__main__":
    df = pd.read_csv("data/wfp_food_prices_ken.csv")
    commodity = "Beans"
    market = "Mombasa"
    unit = "KG"
    pricetype = "Wholesale"

    df["date"] = pd.to_datetime(df["date"], errors="coerce")
    df["price"] = pd.to_numeric(df["price"], errors="coerce")
    df = df.dropna(subset=["date", "price"])

    print(df.dtypes)
    print(df.head(10))
```

```python
valid_df_one = visualize.commodity_date_price_change(df,
                                      commodity,
                                      market,
                                      unit,
                                      pricetype)

valid_df_two = visualize.commodity_market_price_change(df,
                                               commodity,
                                               unit,
                                               pricetype)

print(valid_df_one.head())
print(valid_df_two.head())
```

```python
visualize.visualize_date_commodity_price_change(valid_df_one)
visualize.visualize_market_commodity_price_change(valid_df_two)
```
